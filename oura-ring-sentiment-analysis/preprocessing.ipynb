```python
import pandas as pd
import re
import nltk
```

```python
df = pd.read_csv("oura_ring4_all_reviews.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df["body"] = df["body"].astype(str)
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)   # remove punctuation
    text = re.sub(r"\d+", "", text)       # remove numbers
    text = re.sub(r"\s+", " ", text).strip()  # remove extra spaces
    return text
```

```python
df["clean_body"] = df["body"].apply(clean_text)
```

```python
df[["body", "clean_body"]].head(10)
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

print(sorted(stop_words))
```

```python
important_words = {"not", "no", "never", "should", "need"} #Words we want to keep
stop_words = stop_words - important_words #Removing words we want to keep from stopword list
print("not" in stop_words)
print("no" in stop_words)
print("never" in stop_words)
print("should" in stop_words)
print("need" in stop_words)
```

```python
def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in stop_words])

df["clean_body"] = df["clean_body"].apply(remove_stopwords)
df[["body", "clean_body"]].head(10)
```

```python
df.to_csv("oura_ring4_all_reviews_preprocessed.csv", index=False)
```
